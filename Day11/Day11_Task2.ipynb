```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print("Repeated OrderIDs:", df["OrderID"].duplicated().sum())

print(df[df["OrderID"].duplicated(keep=False)])
```

```python
string_columns = ["Product", "Category"]

for col in string_columns:
    df[col] = df[col].str.strip().str.title()

print(df[string_columns])
```

```python
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

print(df[["OrderID", "OrderDate"]])
print("Date data type:", df["OrderDate"].dtype)
```

```python
invalid_dates = df[df["OrderDate"].isna()]

print(invalid_dates[["OrderID", "OrderDate"]])
```

```python
numeric_columns = ["Quantity", "UnitPrice", "Discount"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
for col in numeric_columns:
    original_values = df_original.loc[df.index, col]

    converted_values = df[col]

    invalid_mask = (
        original_values.notna()
        & converted_values.isna()
    )

    print(f"\nInvalid values in {col}:")
    print(original_values[invalid_mask])
```

```python
print("Quantity:", df["Quantity"].dtype)
print("UnitPrice:", df["UnitPrice"].dtype)
print("Discount:", df["Discount"].dtype)
print("OrderDate:", df["OrderDate"].dtype)
```

```python
df["Product"] = df["Product"].fillna("Unknown")

quantity_median = df["Quantity"].median()

df["Quantity"] = df["Quantity"].fillna(quantity_median)

print(df.isnull().sum())
```

```python
Q1 = df["UnitPrice"].quantile(0.25)
Q3 = df["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outlier_mask = (
    (df["UnitPrice"] < lower_bound)
    | (df["UnitPrice"] > upper_bound)
)

outliers = df.loc[
    outlier_mask,
    ["OrderID", "Product", "UnitPrice"]
]

print(outliers)
```

```python
df["TotalSales"] = df["Quantity"] * df["UnitPrice"]

print(df[[
    "OrderID", "Product", "Quantity",
    "UnitPrice", "TotalSales"
]])
```

```python
df["FinalAmount"] = df["TotalSales"] - df["Discount"]

print(df[[
    "OrderID", "TotalSales", "Discount", "FinalAmount"
]])
```

```python
def categorize_price(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 1000:
        return "Low"
    elif price <= 25000:
        return "Medium"
    else:
        return "High"
```

```python
df["PriceCategory"] = df["UnitPrice"].apply(categorize_price)

print(df[["Product", "UnitPrice", "PriceCategory"]])
```

```python
expensive_products = df[df["UnitPrice"] > 25000]

print(expensive_products)
```
