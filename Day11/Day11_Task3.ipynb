```python
required_columns = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "Country"
]

df = df.dropna(subset=required_columns).copy()

print("Shape after handling missing values:", df.shape)
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(
    "Repeated invoice numbers:",
    df["InvoiceNo"].duplicated().sum()
)
```

```python
df = df.drop_duplicates().copy()

print("Duplicates after cleaning:", df.duplicated().sum())
print("Current shape:", df.shape)
```

```python
df["Description"] = (
    df["Description"]
    .astype("string")
    .str.strip()
)

df["Country"] = (
    df["Country"]
    .astype("string")
    .str.strip()
)

print(df[["Description", "Country"]].head())
```

```python
df["Description"] = (
    df["Description"]
    .astype("string")
    .str.strip()
)

df["Country"] = (
    df["Country"]
    .astype("string")
    .str.strip()
)

print(df[["Description", "Country"]].head())
```

```python
df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"], errors="coerce"
)

df["CustomerID"] = pd.to_numeric(
    df["CustomerID"], errors="coerce"
)

print(df[["Quantity", "UnitPrice", "CustomerID"]].dtypes)
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"], errors="coerce"
)

print("InvoiceDate data type:", df["InvoiceDate"].dtype)
```

```python
print("Missing quantities:", df["Quantity"].isna().sum())
print("Missing prices:", df["UnitPrice"].isna().sum())
print("Invalid dates:", df["InvoiceDate"].isna().sum())
```

```python
df = df.dropna(
    subset=["Quantity", "UnitPrice", "InvoiceDate"]
).copy()

df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

print("Rows after sales cleaning:", len(df))
print(df.head())
```

```python
highest_index = df["UnitPrice"].idxmax()

highest_product = df.loc[
    highest_index,
    ["StockCode", "Description", "UnitPrice"]
]

print("Highest-priced product:")
print(highest_product)
```

```python
df.nlargest(5, "UnitPrice")[
    ["StockCode", "Description", "UnitPrice"]
]
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(df[[
    "InvoiceNo",
    "Description",
    "Quantity",
    "UnitPrice",
    "Revenue"
]].head(10))
```

```python
country_revenue = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(country_revenue)
```

```python
print("Highest-revenue country:", country_revenue.idxmax())

print(
    "Highest revenue (£):",
    round(country_revenue.max(), 2)
)
```

```python
country_revenue.head(5).plot(
    kind="bar",
    figsize=(8, 5),
    title="Top 5 Countries by Revenue",
    ylabel="Revenue (£)",
    xlabel="Country"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_revenue = (
    df.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(product_revenue.head(5))
```
