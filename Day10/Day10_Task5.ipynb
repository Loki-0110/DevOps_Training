```python
import numpy as np
```

```python
print(np.__version__)
```

```python
attendance = np.array([
    7.5, 6.0, 8.2, 5.5, 9.1,
    4.8, 6.8, 8.7, 7.2, 5.9,
    8.0, 9.4, 6.4, 7.8, 8.9,
    5.7, 7.6, 6.9, 9.0, 6.2
])

print(attendance)
```

```python
print("Array:", attendance)
print("Number of dimensions:", attendance.ndim)
print("Shape:", attendance.shape)
print("Size:", attendance.size)
print("Data type:", attendance.dtype)
```

```python
print("Minimum attendance:", np.min(attendance))
print("Maximum attendance:", np.max(attendance))
print("Average attendance:", np.mean(attendance))
print("Total attendance values:", np.sum(attendance))
print("Median attendance:", np.median(attendance))
print("Standard deviation:", np.std(attendance))
print("Variance:", np.var(attendance))
```

```python
print("Sorted attendance:", np.sort(attendance))
print("First value:", attendance[0])
print("Last value:", attendance[-1])
print("First five values:", attendance[:5])
print("Attendance above 8:", attendance[attendance > 8])
```

```python
print("Add 1 to every value:", attendance + 1)
print("Subtract 1:", attendance - 1)
print("Multiply by 2:", attendance * 2)
print("Divide by 2:", attendance / 2)
```

```python
categories = np.where(attendance >= 8, "High", "Low")

print(categories)
print("High attendance count:", np.sum(attendance >= 8))
print("Low attendance count:", np.sum(attendance < 8))
```
